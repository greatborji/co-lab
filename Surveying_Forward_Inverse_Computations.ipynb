# Surveying Computations: Forward and Inverse Problems

A beginner-friendly notebook for surveying students. After uploading this notebook to GitHub, students can open it directly in Google Colab.

## Coordinate convention
- Coordinates are Easting (E) and Northing (N).
- Azimuth is measured clockwise from North (0° to less than 360°).
- Use consistent units for coordinates and distance, such as metres.

Forward: E2 = E1 + D sin(Az), N2 = N1 + D cos(Az).

Inverse: D = sqrt((E2-E1)^2 + (N2-N1)^2). Azimuth is computed clockwise from North.


```python
import math

def forward_computation(E1, N1, distance, azimuth_deg):
    """Compute point 2 from point 1, distance, and azimuth clockwise from North."""
    if distance < 0:
        raise ValueError("Distance must be non-negative.")
    az = math.radians(azimuth_deg % 360)
    E2 = E1 + distance * math.sin(az)
    N2 = N1 + distance * math.cos(az)
    return E2, N2

def inverse_computation(E1, N1, E2, N2):
    """Compute horizontal distance and azimuth clockwise from North."""
    dE, dN = E2 - E1, N2 - N1
    distance = math.hypot(dE, dN)
    if distance == 0:
        return 0.0, None  # Azimuth is undefined for coincident points.
    azimuth_deg = math.degrees(math.atan2(dE, dN)) % 360
    return distance, azimuth_deg

print('Functions are ready.')
```

## Interactive calculator
Run this cell and select Forward or Inverse. Enter azimuth in decimal degrees.


```python
print('Surveying Computations')
print('1 - Forward computation')
print('2 - Inverse computation')
choice = input('Choose calculation type (1 or 2): ').strip()

if choice == '1':
    print('\nFORWARD COMPUTATION')
    E1 = float(input('Enter starting point Easting (E1): '))
    N1 = float(input('Enter starting point Northing (N1): '))
    distance = float(input('Enter distance to the next point: '))
    azimuth = float(input('Enter azimuth in degrees clockwise from North: '))
    E2, N2 = forward_computation(E1, N1, distance, azimuth)
    print('\nRESULT')
    print(f'Point 2 Easting (E2):  {E2:.4f}')
    print(f'Point 2 Northing (N2): {N2:.4f}')

elif choice == '2':
    print('\nINVERSE COMPUTATION')
    E1 = float(input('Enter point 1 Easting (E1): '))
    N1 = float(input('Enter point 1 Northing (N1): '))
    E2 = float(input('Enter point 2 Easting (E2): '))
    N2 = float(input('Enter point 2 Northing (N2): '))
    distance, azimuth = inverse_computation(E1, N1, E2, N2)
    print('\nRESULT')
    print(f'Horizontal distance: {distance:.4f}')
    if azimuth is None:
        print('Azimuth: undefined (the two points have identical coordinates).')
    else:
        print(f'Azimuth clockwise from North: {azimuth:.4f} degrees')

else:
    print('Invalid choice. Run this cell again and enter 1 or 2.')
```
