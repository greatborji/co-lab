```python
import numpy as np
import matplotlib.pyplot as plt

dem = np.random.randint(100,500,(40,40))
plt.imshow(dem, cmap="terrain")
row = dem[1,  :]
fig, axs = plt.subplots(2,1 , figsize =(10,10))
axs[0].imshow(dem,cmap="terrain")
#axs[0].set_title("terrain elevation map")
axs[0].grid(True)
plt.colorbar(axs[0].images[0],label="elevation(m)")
axs[1].plot(row,marker="o")
axs[1].set_title("row 2 profile")
axs[1].set_xlabel("station")
axs[1].set_ylabel("elevation")
axs[1].grid(True)
plt.show()
```
